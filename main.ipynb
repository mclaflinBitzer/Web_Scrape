```python
import pandas as pd
import logging
from Scripts.nr_scrape import nr_fetch_method, nr_url_extraction, nr_article_scraping_method
from Scripts.trane_scrape import trane_fetch_method, trane_url_extraction, trane_scrape_article
from Scripts.API_push import api_push
from Scripts.logger import setup_logger
from dotenv import load_dotenv
import os
from datetime import datetime
```

```python
logger = setup_logger(__name__)
```

```python
logger.info("Starting the web scraping pipeline.")
```

#### base urls

```python
## Base URLs
load_dotenv()
nr_url = os.getenv("nr_base_url")
trane_url = os.getenv("trane_base_url")
```

#### list creation for storing article data in list - dict format prior to dataframe creation

```python
all_articles = []
```

#### Natural Refrigerants Scrape Process

```python
## fetching html from base natural refrigerants url
logger.info(f"fetching base natural refrigerants html from url: {nr_url}")
nr_soup = nr_fetch_method(nr_url)

logger.info("Extracting sub article links from base NR html")
nr_article_links = nr_url_extraction(nr_soup)

for link in nr_article_links:
    logger.info(f"Scraping article data from link: {link}")
    all_articles.append(nr_article_scraping_method(link))
```

#### Trane Technologies Scrape Process

```python
## scraping html from base Trane url
logger.info(f"fetching base Trane html from url: {trane_url}")
trane_soup = trane_fetch_method(trane_url)

## extracting sub article links from base Trane html    
logger.info("Extracting sub article links from base Trane html")
trane_article_links = trane_url_extraction(trane_soup)

for link in trane_article_links:
    logger.info(f"Scraping article data from link: {link}")
    all_articles.append(trane_scrape_article(link))
```

### Writing to Excel File

```python
logger.info("Web scraping and data compilation complete. saving compiled data to excel file.")
compiled_article_data = pd.DataFrame(all_articles)
compiled_article_data.to_excel(os.getenv("output_file"), index=False)
```

### Pushing the data to Comintelli API

```python
logger.info("Initiating the api push process")
api_push()
logger.info("API push process complete. Ending the web scraping pipeline.")
```
