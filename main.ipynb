```python
import pandas as pd
from Scripts.nr_scrape import nr_fetch_method, nr_url_extraction, nr_article_scraping_method
from Scripts.trane_scrape import trane_fetch_method, trane_url_extraction, trane_scrape_article
from Scripts.API_push import api_push
```

#### base urls

```python
## Base URLs
nr_url = 'https://naturalrefrigerants.com/news/'
trane_url = 'https://www.tranetechnologies.com/en/index/news.html'
```

#### dataframe creation

```python
## dataframe defining
## columns: title, summary, publication_date->dateline, article_body->newslinetext, attachmenturl
compiled_article_data = pd.DataFrame(columns=['title', 'summary', 'dateline', 'newslinetext', 'attachmenturl','source'])
```

#### Natural Refrigerants Scrape Process

```python
nr_soup = nr_fetch_method(nr_url)
nr_article_links = nr_url_extraction(nr_soup)
for link in nr_article_links:
    compiled_article_data = pd.concat([compiled_article_data, nr_article_scraping_method(link)])
```

#### Trane Technologies Scrape Process

```python
trane_soup = trane_fetch_method(trane_url)

trane_article_links = trane_url_extraction(trane_soup)

for link in trane_article_links:
    compiled_article_data = pd.concat([compiled_article_data, trane_scrape_article(link)], ignore_index=True)
```

### Writing to Excel File

```python
compiled_article_data.to_excel('Data/compiled_article_data.xlsx', index=False)
```
