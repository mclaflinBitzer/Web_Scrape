```python
import pandas as pd
from dotenv import load_dotenv
import os
import requests
```

```python
load_dotenv()

clean_api_url = os.getenv("clean_api_url")
APIid = os.getenv("APIid")
authKey =  os.getenv("authKey")
customerGUID = os.getenv("customerGUID")
accessGroups = os.getenv("accessGroups")

method_api_url = os.getenv("method_api_url")
post_authKey = os.getenv("post_authKey")
post_APIid = os.getenv("post_APIid")
```

```python
df = pd.read_excel('Data/compiled_article_data.xlsx')
```

```python
data = { 
    "APIid": post_APIid, 
    "customerGUID": customerGUID, 
    "authKey": post_authKey,
    "title": str(df['title'].iloc[0]), 
    "summary": str(df['summary'].iloc[0]),
    "body": str(df['newslinetext'].iloc[0]),
    "source": str(df['source'].iloc[0]), 
    "pubdate": df['dateline'].iloc[0].isoformat(),
    "link": df['attachmenturl'].iloc[0], 
    "topicIds": 135575, 
    "accessGroup": accessGroups 
    } 


r = requests.post(method_api_url,data=data, verify=False) 
print(r) 
print(r.content)
```
